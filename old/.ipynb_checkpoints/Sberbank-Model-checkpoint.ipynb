```python
import pandas as pd
import numpy as np
from pathlib import Path

current_directory = str(Path.cwd())
print(current_directory)

data = pd.read_csv(current_directory+"/train.csv")
data = data.drop(columns=["id"])
```

```python
X = data.iloc[:,:-1]
y = data.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
Q1 = y_train.quantile(0.25)
Q3 = y_train.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

mask = (y_train >= lower_bound) & (y_train <= upper_bound)

X_train = X_train.loc[mask]
y_train = y_train.loc[mask]

# 3. Sampling (npr. 10000 redova)
X_train = X_train.sample(n=10000, random_state=42)
y_train = y_train.loc[X_train.index]

# y_train_class, bins = pd.qcut(
#     y_train, q=10, labels=False, retbins=True, duplicates="drop"
# )
# y_test_class = pd.cut(
#     y_test, bins=bins, labels=False, include_lowest=True
# )
```

```python
X_train["year"] = X_train["timestamp"].dt.year
X_train["month"] = X_train["timestamp"].dt.month
X_train["quarter"] = X_train["timestamp"].dt.quarter

X_test["year"] = X_test["timestamp"].dt.year
X_test["month"] = X_test["timestamp"].dt.month
X_test["quarter"] = X_test["timestamp"].dt.quarter
```

```python
# Napravi kopiju da ne menjaš originalni X_train
train_with_target = X_train.copy()

# Dodaj target kao novu kolonu
train_with_target["price_doc"] = y_train.values
```

```python
# 1. Dodaj price_per_sq sa bezbednim deljenjem
train_with_target["price_per_sq"] = (
    train_with_target["price_doc"] / train_with_target["full_sq"]
).replace([np.inf, -np.inf], np.nan)

# 2. Grupisi po sub_area + year + quarter
grouped_sq = (
    train_with_target
    .groupby(["sub_area", "year", "quarter"], observed=True)["price_per_sq"]
    .mean()
    .reset_index()
    .rename(columns={"price_per_sq": "avg_price_per_sq_subarea"})
)

# 3. Dodaj lag
grouped_sq["lag_price_per_sq_subarea"] = (
    grouped_sq.groupby("sub_area", observed=True)["avg_price_per_sq_subarea"].shift(1)
)

# 4. Pre merge-a obriši stare kolone ako postoje
for df in [X_train, X_test]:
    df.drop(
        columns=[c for c in df.columns if "avg_price_per_sq_subarea" in c or "lag_price_per_sq_subarea" in c],
        inplace=True,
        errors="ignore"
    )

# 5. Merge nazad
X_train = X_train.merge(grouped_sq, on=["sub_area", "year", "quarter"], how="left")
X_test = X_test.merge(grouped_sq, on=["sub_area", "year", "quarter"], how="left")

# 6. Osiguraj da nema duplikata kolona (_x, _y itd.)
X_train = X_train.loc[:, ~X_train.columns.duplicated()]
X_test = X_test.loc[:, ~X_test.columns.duplicated()]

# 7. Provera novih kolona
target_cols = ["avg_price_per_sq_subarea", "lag_price_per_sq_subarea"]

print("Naše nove kolone:", [c for c in target_cols if c in X_train.columns])
print(X_train[target_cols].head())

```

#Data Preprocesing

```python
from sklearn.base import BaseEstimator, TransformerMixin

class FrequencyEncoder(BaseEstimator, TransformerMixin):
    def __init__(self, columns=None, normalize=True):
        self.columns = columns
        self.normalize = normalize
        self.freq_maps = {}

    def fit(self, X, y=None):
        X = pd.DataFrame(X, columns=self.columns if not isinstance(X, pd.DataFrame) else X.columns)
        for col in self.columns:
            self.freq_maps[col] = X[col].value_counts(normalize=self.normalize)
        return self

    def transform(self, X):
        X = pd.DataFrame(X, columns=self.columns if not isinstance(X, pd.DataFrame) else X.columns)
        X_trans = X.copy()
        for col in self.columns:
            X_trans[col] = X[col].map(self.freq_maps[col]).fillna(0)
        return X_trans

    def get_feature_names_out(self, input_features=None):
        return [col + "_freq" for col in self.columns]
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

    def fit(self, X, y):
        X = pd.DataFrame(X, columns=X.columns if hasattr(X, "columns") else None)

        # 1️⃣ Izbaci kolone sa previše NaN
        cols_to_drop = []
        for col in X.columns:
            missing_ratio = X[col].isnull().mean()
            if missing_ratio > self.missing_threshold:
                cols_to_drop.append(col)

        X_filtered = X.drop(columns=cols_to_drop, errors="ignore")

        # 2️⃣ Selekcija po korelaciji sa targetom
        target_corr = X_filtered.corrwith(y).abs()
        selected_features = target_corr[target_corr > self.corr_threshold].dropna().index.tolist()

        if not selected_features:
            self.cols_to_keep_ = []
            self.imputer_ = SimpleImputer(strategy="median")
            self.imputer_.fit(np.array([[0]]))
            print("⚠️ Nema numeričkih kolona iznad praga korelacije.")
            return self

        df_filtered = X_filtered[selected_features]

        # 3️⃣ Ukloni prejako korelisane kolone
        corr_matrix = df_filtered.corr().abs()
        high_corr_pairs = [
            (c1, c2)
            for c1 in corr_matrix.columns
            for c2 in corr_matrix.columns
            if c1 < c2 and corr_matrix.loc[c1, c2] > self.pairs_threshold
        ]

        to_drop = []
        for c1, c2 in high_corr_pairs:
            if target_corr.get(c1, 0) < target_corr.get(c2, 0):
                to_drop.append(c1)
            else:
                to_drop.append(c2)

        self.cols_to_keep_ = [c for c in selected_features if c not in to_drop]

        # 4️⃣ Fit imputer
        if self.cols_to_keep_:
            self.imputer_ = SimpleImputer(strategy="median")
            self.imputer_.fit(df_filtered[self.cols_to_keep_])
        else:
            self.imputer_ = SimpleImputer(strategy="median")
            self.imputer_.fit(np.array([[0]]))

        print("✅ Numeričke kolone koje ćemo zadržati:", self.cols_to_keep_ if self.cols_to_keep_ else "nema")
        return self




```

```python
from sklearn.base import BaseEstimator, TransformerMixin

class DateExtractor(BaseEstimator, TransformerMixin):
    def __init__(self, col="timestamp"):
        self.col = col
    
    def fit(self, X, y=None):
        return self
    
    def transform(self, X):
        X_ = X.copy()
        X_[self.col] = pd.to_datetime(X_[self.col])
        X_["year"] = X_[self.col].dt.year
        X_["month"] = X_[self.col].dt.month
        X_["quarter"] = X_[self.col].dt.quarter
        X_ = X_.drop(columns=[self.col])  # izbaci originalni timestamp
        return X_
```

```python
# ---------------------------
# Test Data
# ---------------------------
# test_df = pd.DataFrame({
#     "gender": ["male", "female", "male", "female", "male"],
#     "city": ["Belgrade", "Novi Sad", "Belgrade", "Niš", "Belgrade"],
#     "ecology": ["excellent", "good", "poor", "no data", "good"],
#     "quality": ["excellent", "satisfactory", "poor", "good", "no data"]
# })
# ---------------------------
# Column groups
# ---------------------------
nominal_cols = [
    "culture_objects_top_25","thermal_power_plant_raion","incineration_raion",
    "oil_chemistry_raion","radiation_raion","railroad_terminal_raion",
    "big_market_raion","nuclear_reactor_raion","detention_facility_raion",
    "water_1line","big_road1_1line","railroad_1line","product_type"
]

ordinal_cols = ["ecology"]
freq_cols = ["sub_area"]

# ---------------------------
# Categories for ordinal cols
# ---------------------------
ordinal_categories = {
    "ecology": ["excellent", "good", "satisfactory", "poor"]
}

ord_categories = [ordinal_categories[col] for col in ordinal_cols]

# ---------------------------
# Preprocessor
# ---------------------------
categorical_preprocessor = ColumnTransformer(
    transformers=[
        # Ordinal
        ("ord", Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("ord", OrdinalEncoder(
                categories=ord_categories,
                handle_unknown="use_encoded_value", unknown_value=-1
            ))
        ]), ordinal_cols),

        # Frequency
        ("freq", Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("freq", FrequencyEncoder(columns=freq_cols, normalize=True))
        ]), freq_cols),

        # OneHot
        ("ohe", Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("ohe", OneHotEncoder(drop="if_binary", sparse_output=False, handle_unknown="ignore"))
        ]), nominal_cols)
    ],
    remainder="drop"
)

# ---------------------------
# Transform
# ---------------------------
transformed = categorical_preprocessor.fit_transform(X_train)
```

```python
from sklearn.pipeline import Pipeline

# 2. pipeline koji spaja categorical + numeric (bez obrade numeričkih)
cat_num_merger = ColumnTransformer(
    transformers=[
        ("cat", categorical_preprocessor, nominal_cols + ordinal_cols + freq_cols),
        ("num", SimpleImputer(strategy="median"), numeric_cols)
    ]
)
# Numerički preprocessor
numeric_preprocessor = NumericFeatureSelector(
    missing_threshold=0.3,
    corr_threshold=0.28,
    pairs_threshold=0.85
)
# 1️⃣ Definiši ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        # kategorijske kroz tvoj categorical_preprocessor
        ("cat", categorical_preprocessor, nominal_cols + ordinal_cols + freq_cols),

        # numeričke kroz NumericFeatureSelector
        ("num", NumericFeatureSelector(
            missing_threshold=0.3,
            corr_threshold=0.05,
            pairs_threshold=0.95
        ), numeric_cols)
    ]
)

# 2️⃣ Full pipeline: preprocesiranje + model
pipeline = Pipeline(steps=[
    ("date", DateExtractor("timestamp")),
    ("preprocessor", preprocessor),
    ("model", XGBRegressor(
        n_estimators=1000,
        learning_rate=0.05,
        max_depth=8,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    ))
])
```

```python
import xgboost as xgb
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_squared_error, mean_absolute_error, median_absolute_error,
    r2_score, explained_variance_score, mean_squared_log_error
)

# 1️⃣ Pipeline
pipeline = Pipeline(steps=[
    ("preprocessor", cat_num_merger),  # koristi tvoj ColumnTransformer
    ("model", xgb.XGBRegressor(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        tree_method="hist",
        random_state=42
    ))
])

# 2️⃣ Fit
pipeline.fit(X_train, y_train)

# 3️⃣ Predikcija
y_pred = pipeline.predict(X_test)

# 4️⃣ Feature names (posle preprocesora)
feature_names = pipeline.named_steps["preprocessor"].get_feature_names_out()

# 5️⃣ Feature importance
xgb_model = pipeline.named_steps["model"]
booster = xgb_model.get_booster()
booster.feature_names = feature_names.tolist()

importance_dict = booster.get_score(importance_type="gain")
importance_df = pd.DataFrame(
    importance_dict.items(),
    columns=["Feature", "Importance"]
).sort_values(by="Importance", ascending=False).head(20)

print("\n📊 Top 20 feature importance (gain):")
print(importance_df)

# 6️⃣ Evaluacija
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
medae = median_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
evs = explained_variance_score(y_test, y_pred)
mape = np.mean(np.abs((y_test - y_pred) / y_test)) * 100
rmsle = np.sqrt(mean_squared_log_error(y_test, np.maximum(y_pred, 0)))

print("\n📈 Metrike:")
print("RMSLE:", rmsle)
print("MSE:", mse)
print("RMSE:", rmse)
print("MAE:", mae)
print("Median AE:", medae)
print("R²:", r2)
print("Explained Variance:", evs)
print("MAPE:", mape, "%")

# 7️⃣ Plot sa pravim imenima
plt.figure(figsize=(10, 6))
plt.barh(importance_df["Feature"], importance_df["Importance"], color="skyblue")
plt.gca().invert_yaxis()
plt.title("Top 20 Feature Importance (gain)", fontsize=14)
plt.xlabel("Importance (gain)")
plt.show()

```

```python
from itertools import combinations
import numpy as np
from sklearn.pipeline import Pipeline
import xgboost as xgb
from sklearn.metrics import mean_squared_log_error

results = []

def eval_subset(extra_cols):
    # napravi novu listu numeric_cols tako da sadrži osnovne + dodatne
    used_num_cols = [c for c in numeric_cols if c not in new_cols] + list(extra_cols)
    
    # 1️⃣ Definiši ColumnTransformer
    preprocessor = ColumnTransformer(
        transformers=[
            # kategorijske kroz tvoj categorical_preprocessor
            ("cat", categorical_preprocessor, nominal_cols + ordinal_cols + freq_cols),
    
            # numeričke kroz NumericFeatureSelector
            ("num", NumericFeatureSelector(
                missing_threshold=0.3,
                corr_threshold=0.05,
                pairs_threshold=0.95
            ), numeric_cols)
        ]
    )

    pipe = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", xgb.XGBRegressor(
            n_estimators=2000,        
            learning_rate=0.03,       
            max_depth=8,
            subsample=0.8,
            colsample_bytree=0.8,
            colsample_bylevel=0.8,
            colsample_bynode=0.8,
            min_child_weight=5,
            gamma=1,
            reg_alpha=0.1,
            reg_lambda=5,
            booster="dart",          
            tree_method="hist",       
            random_state=42,
            n_jobs=-1
        ))
    ])

    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)

    return np.sqrt(mean_squared_log_error(y_test, np.maximum(pred, 0)))

# 1️⃣ base model bez dodatnih kolona
rmsle_base = eval_subset([])
results.append((("bez novih kolona",), rmsle_base))
print("Bez novih kolona -> RMSLE:", rmsle_base)

# 2️⃣ testiraj sve kombinacije novih kolona
new_cols = [
    "year", "month", "quarter", "months_since_start",
    "lag_price_per_sq_subarea", "floor_ratio"
]

for r in range(1, len(new_cols)+1):
    for subset in combinations(new_cols, r):
        rmsle = eval_subset(subset)
        results.append((subset, rmsle))
        print(f"Subset {subset}: RMSLE={rmsle:.4f}")

# 3️⃣ sortiraj
results_sorted = sorted(results, key=lambda x: x[1])
print("\nTop 5 najboljih kombinacija:")
for subset, score in results_sorted[:5]:
    print(f"{subset} -> RMSLE={score:.4f}")

```
